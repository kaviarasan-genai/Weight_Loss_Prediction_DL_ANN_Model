```python
# 🚀 Project 1 : Lifestyle-Based Weight Loss Prediction with ANN

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

data = pd.read_csv("weight_loss_dataset.csv")
df1 = pd.DataFrame(data)
df = df1.copy()
df
```

```python
# Preprocessing EDA Method

print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

```python
# Column Encode Categorical Columns Into Numerical Columns

df["Weight Loss or Gain"] = df["Weight Loss or Gain"].map({"Weight Loss": 0,
                                                           "Weight Gain": 1})
print(df.head())
```

```python
# Taking Model Features or Nodes

x = df[["Diet Score", "Sleep Hours", "Exercise Hours", "Stress Level"]].values
y = df["Weight Loss or Gain"].values
```

```python
# Processing on Model Train Test Split | Import model

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.20, random_state = 42)

model = Sequential()

model.add(Dense(4, activation = "relu", input_shape = (4, )))
model.add(Dense(1, activation = "sigmoid"))

model.summary()
```

```python
# Compile Model

model.compile(optimizer = "adam", loss = "binary_crossentropy", metrics = ["accuracy"])
```

```python
# Fit Model | Epochs

model.fit(x_train,y_train, epochs=100)
```

```python
# Model Evaluate | Accuracy Testing

model.evaluate(x_test,y_test)
```

```python
# Model Testing New Datas

def new_input(datas):
  while True:
    try:
      return float(input(datas))
    except ValueError:
      print("Invalid Inputs!!! Please Use Numbers Only.")

Diet_Score     = new_input("Enter Your Diet Score     : ")
Sleep_Hours    = new_input("Enter Your Sleeping Hours : ")
Exercise_Hours = new_input("Enter Your Exercise Hours : ")
Stress_Level   = new_input("Enter Your Stress Level   : ")

new_data = np.array([[Diet_Score, Sleep_Hours, Exercise_Hours, Stress_Level]])

prediction = model.predict(new_data)

if prediction < 0.5 :
  print("_"*35)
  print("""   If You Maintain This Routine
     You Will Lose Weight Quickly""")
  print("~"*35)
else :
  print("_"*36)
  print("""  If You Continually Routine This
      You Will Gain More Weight """)
  print("~"*36)
```
